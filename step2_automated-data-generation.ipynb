```python
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
fake = Faker("vi_VN")
Faker.seed(SEED)
n=1000

df = pd.read_csv("pharmacity_personal_care.csv").drop(columns=["URL"])

products = df.sample(n=n,replace=True,random_state=42).reset_index(drop=True)

customer_ids = [f"C{i:04d}"for i in range(1, n + 1)]

customer_names = [fake.name()for _ in range(n)]

ages = np.random.normal(loc=30,scale=10,size=n).round().astype(int)

ages = np.clip(ages, 18, 70)

genders = random.choices(['Nam', 'Nữ'],weights=[0.45, 0.55],k=n)

cities = random.choices(['TP. Hồ Chí Minh','Hà Nội','Đà Nẵng','Cần Thơ','Hải Phòng'], weights=[0.40, 0.30, 0.10, 0.10, 0.10],k=n)
payment_methods = random.choices(['Tiền mặt','Thẻ ngân hàng','Ví điện tử','COD'],weights=[0.20, 0.25, 0.25, 0.30],k=n)

quantities = np.random.choice([1, 2, 3, 4, 5],size=n,p=[0.50, 0.25, 0.15, 0.07, 0.03])

purchase_dates = pd.to_datetime(np.random.choice(pd.date_range('2026-01-01','2026-09-30'),size=n))

total_amount = (products['Price'].values * quantities)

Order_df = pd.DataFrame({'OrderID': [f'ORD{i:05d}'for i in range(1, n + 1)],
'CustomerID': customer_ids,
'CustomerName': customer_names,
'Age': ages,
'Gender': genders,
'ProductID': products['ProductID'].values,
'ProductName': products['ProductName'].values,
'Category': products['Category'].values,
'Brand': products['Brand'].values,
'Price': products['Price'].values,
'Quantity': quantities,
'TotalAmount': total_amount,
'PurchaseDate': purchase_dates,
'City': cities,
'PaymentMethod': payment_methods})
```

```python
print("Kích thước dataset:", Order_df.shape)

display(Order_df.head(5))
```

```python
Order_df.to_csv('pharmacity_simulated_transactions.csv',index=False,encoding='utf-8-sig')

print("Đã xuất file:")
print("pharmacity_simulated_transactions.csv")
```
